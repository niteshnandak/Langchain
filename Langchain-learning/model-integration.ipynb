### Model Integration with OpenAI, Gemini and Groq


```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
os.environ["NVIDIA_API_KEY"] = os.getenv("NVIDIA_API_KEY")
```

### Nemotron Model Integration

```python
from langchain.chat_models import init_chat_model
model = init_chat_model("nvidia/nemotron-3-ultra-550b-a55b", model_provider="nvidia")
model
```

```python
# Invoking the model 

res = model.invoke("How are you")
res.content
```

### Gemini Model Integration

```python
from langchain.chat_models import init_chat_model
from IPython.display import display, Markdown

model = init_chat_model("google_genai:gemini-3.5-flash-lite")
response = model.invoke("How do Parrots talk")
Markdown(response.content[0]["text"])
```

```python
# ChatGoogleGemini
from langchain_google_genai import ChatGoogleGenerativeAI
from IPython.display import display, Markdown

model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
response = model.invoke("Egg is veg or non-veg?")
Markdown(response.content[0]["text"])
```

### Groq Model Integration

```python
from langchain.chat_models import init_chat_model
from IPython.display import display, Markdown


model = init_chat_model("qwen/qwen3.8-27b", model_provider="groq")
response = model.invoke("Power figures of Mahindra XUV 700")
Markdown(response.content)
```

```python
# ChatGroq
from langchain_groq import ChatGroq
from IPython.display import display, Markdown

model = ChatGroq(model="qwen/qwen3.8-27b")
response = model.invoke("What is the latest Iphone model")
Markdown(response.content)
```
