### Tools

##### Models can request to call tools that perform tasks such as fetch data from DB, searching web pages, calling an API or running code. 


```python
import os 
from langchain.chat_models import init_chat_model
from IPython.display import display, Markdown

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("qwen/qwen3.8-27b",reasoning_effort="medium",
    reasoning_format="parsed", model_provider="groq")
response = model.invoke("Power figures of Mahindra XUV 700")
Markdown(response.content)
```

```python
from langchain.tools import tool

# Whenever we use this decorator on top of a function it automatically becomes a tool
@tool
def get_weather(location:str)->str:
    """Get the weather of a location"""
    return f"It's sunny at {location}"

tool_models = model.bind_tools([get_weather])
```

```python
response = tool_models.invoke("what's the weather at Boston ?")
print(response)
for tool_call in response.tool_calls:
    print(f"Tool: {tool_call['name']}")
    print(f"Args: {tool_call['args']}")
```

### Tool Execution Loop

```python
# 1. Model generates tool calls
msg = [{"role": "user", "content":"What's the weather at California?"}]
ai_msg = tool_models.invoke(msg)
msg.append(ai_msg)

# 2. Execute tool and collect results
for tool_call in ai_msg.tool_calls:
    tool_result = get_weather.invoke(tool_call)
    msg.append(tool_result)

# 3. Pass the result back to the model for final response
final_msg = tool_models.invoke(msg)
print(final_msg.text)
```
